```python
#1.IMPORTING LIBRARIES

import pandas as pd
# import → Python keyword that brings the specified module into the current program
# pandas → Python library used for data manipulation and analysis, especially with tables (DataFrames)
# as → Python keyword used to give an imported item a shorter alias/nickname
# pd → The alias assigned to pandas, used to call the library more easily in code

import numpy as np
# import → Python keyword that brings the specified module into the current program
# numpy → Python library used for numerical computing and working with arrays
# as → Python keyword used to give an imported item a shorter alias/nickname
# np → The alias assigned to numpy, used to call the library more easily in code

import itertools
# import → Python keyword that brings the specified module into the current program
# itertools → Built-in Python module that provides tools for creating and working with iterators (e.g. combinations, loops)

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
# from → Python keyword used to import specific parts (not the whole module) from a library
# sklearn.feature_extraction.text → Module in scikit-learn for converting text into numerical features
# import → Brings the specified classes into the current program
# CountVectorizer → Class that converts text into a matrix of word counts
# TfidfVectorizer → Class that converts text into a matrix of TF-IDF weighted scores
# HashingVectorizer → Class that converts text into a fixed-size numerical matrix using hashing

from sklearn.model_selection import train_test_split
# from → Python keyword used to import specific parts from a library
# sklearn.model_selection → Module in scikit-learn for splitting and validating datasets
# import → Brings the specified function into the current program
# train_test_split → Function that splits data into training and testing sets

from sklearn.linear_model import PassiveAggressiveClassifier
# from → Python keyword used to import specific parts from a library
# sklearn.linear_model → Module in scikit-learn containing linear model algorithms
# import → Brings the specified class into the current program
# PassiveAggressiveClassifier → Classification algorithm that updates aggressively on misclassified examples

from sklearn.naive_bayes import MultinomialNB
# from → Python keyword used to import specific parts from a library
# sklearn.naive_bayes → Module in scikit-learn containing Naive Bayes algorithms
# import → Brings the specified class into the current program
# MultinomialNB → Naive Bayes classifier suited for text/word-count data

from sklearn import metrics
# from → Python keyword used to import specific parts from a library
# sklearn → scikit-learn library, used for machine learning
# import → Brings the specified module into the current program
# metrics → Module containing tools to evaluate model performance (accuracy, confusion matrix, etc.)

import matplotlib.pyplot as plt
# import → Python keyword that brings the specified module into the current program
# matplotlib.pyplot → Module used for creating plots and visualizations
# as → Python keyword used to give an imported item a shorter alias/nickname
# plt → The alias assigned to matplotlib.pyplot, used to call plotting functions more easily

# This cell imports all the libraries needed for text vectorization, model training (PassiveAggressiveClassifier, MultinomialNB), 
# evaluation (metrics), and visualization (plt).
```

```python
#2.LOADING THE DATASET

df = pd.read_csv('news.csv', index_col=None)
# df → Variable storing the loaded dataset
# pd.read_csv → Function that reads a CSV file and loads it into a DataFrame
# 'news.csv' → Name of the file being read
# index_col=None → Tells pandas not to use any column as the row index (uses default numeric index)

# This cell loads the news.csv file into a DataFrame called df.
```

```python
#3.VIEWING THE DATASET

df
# df → Displays the contents of the df DataFrame (the loaded news.csv data)

# This cell outputs df to inspect the loaded dataset.
```

```python
#4.DROPPING UNWANTED COLUMN

dataset = df.drop("Unnamed: 0",axis=1)
# dataset → Variable storing the cleaned DataFrame
# df.drop → Method that removes a specified column or row from the DataFrame
# "Unnamed: 0" → Name of the column being removed (usually a leftover index column from the CSV)
# axis=1 → Specifies that a column (not a row) should be dropped

# This cell removes the unwanted "Unnamed: 0" column from df and stores the result in dataset.
```

```python
#5.VIEWING THE CLEANED DATASET

dataset
# dataset → Displays the contents of the dataset DataFrame (news data without the "Unnamed: 0" column)

# This cell outputs dataset to inspect the cleaned data.
```

```python
#6.EXTRACTING THE TARGET LABEL

y = dataset["label"]
# y → Variable storing the target/output column
# dataset["label"] → Selects the "label" column from dataset (e.g. REAL/FAKE)

# This cell extracts the "label" column from dataset and stores it in y as the target variable.
```

```python
#7.SPLITTING DATA INTO TRAIN AND TEST SETS

X_train, X_test, y_train, y_test = train_test_split(dataset['text'], y, test_size=0.33, random_state=53)
# X_train, X_test, y_train, y_test → Four variables that will receive the split data
# = → Assignment operator; unpacks the results into these variables
# train_test_split → Function that splits data into training and testing sets
# dataset['text'] → Selects the "text" column from dataset (the input feature)
# y → The target/output column (labels) defined earlier
# test_size=0.33 → Reserves 33% of the data for testing, 67% for training
# random_state=53 → Fixed seed value so the split is the same every time the code runs

# This cell splits the text data and labels into training (67%) and testing (33%) sets.
```

```python
#8.VECTORIZING TEXT USING HASHINGVECTORIZER (NON-NEGATIVE OUTPUT)

hash_vectorizer = HashingVectorizer(stop_words='english', alternate_sign=False)
# hash_vectorizer → Variable storing the vectorizer object
# HashingVectorizer → Class that converts text into a fixed-size numerical matrix using hashing
# stop_words='english' → Removes common English words that add little meaning
# alternate_sign=False → Forces all hashed values to stay non-negative (required for MultinomialNB)

hash_train = hash_vectorizer.fit_transform(X_train)
# hash_train → Variable storing the vectorized training data
# hash_vectorizer.fit_transform → Applies the hashing trick to X_train

hash_test = hash_vectorizer.transform(X_test)
# hash_test → Variable storing the vectorized testing data
# hash_vectorizer.transform → Applies the same hashing to X_test

# This cell re-vectorizes with alternate_sign=False, so hash_train contains no negative values and works with MultinomialNB.
```

```python
#9.INSPECTING HASH_TRAIN OBJECT

dir(hash_train)
# dir → Built-in function that lists all attributes and methods available on an object
# hash_train → The vectorized training data (sparse hashed matrix) whose attributes/methods are being listed

# This cell lists all the methods and properties available on the hash_train sparse matrix object.
```

```python
#10.CHECKING FEATURE MATRIX SIZE

hash_train.shape
# hash_train → The vectorized training data (sparse hashed matrix)
# .shape → Returns the dimensions (rows, columns) of the matrix — columns = fixed hash space size (default 1,048,576)

# This cell shows the shape of hash_train, since HashingVectorizer has no vocabulary to count (unlike Count/TfidfVectorizer).
```

```python
#11.VIEWING A SMALL SLICE OF HASH_TRAIN (SPARSE-SAFE)

print(hash_train[:5].toarray())
# hash_train[:5] → Selects only the first 5 rows instead of the full 4244-row matrix
# .toarray() → Converts just that small slice into a dense array (safe on memory)

# This cell previews a few rows of hash_train without running out of memory.
```

```python
#12.TRAINING AND EVALUATING MULTINOMIAL NAIVE BAYES MODEL (HASHING)

clf = MultinomialNB()
# clf → Variable storing the classifier object
# MultinomialNB() → Creates a new instance of the Multinomial Naive Bayes classifier

clf.fit(hash_train, y_train)
# clf.fit → Method that trains the model
# hash_train → The vectorized training text data (non-negative hashed input)
# y_train → The training labels (output)

pred = clf.predict(hash_test)
# pred → Variable storing the model's predictions
# clf.predict → Method that predicts labels for new/unseen data
# hash_test → The vectorized testing text data (non-negative hashed input)

score = metrics.accuracy_score(y_test, pred)
# score → Variable storing the accuracy result
# metrics.accuracy_score → Function that compares actual vs predicted labels to calculate accuracy
# y_test, pred → Actual labels vs predicted labels being compared

print("accuracy:   %0.3f" % score)
# print → Built-in function that displays output
# "accuracy:   %0.3f" → Text label with a placeholder for a number formatted to 3 decimal places
# % score → Inserts the score value into the placeholder

cm = metrics.confusion_matrix(y_test, pred, labels=['FAKE', 'REAL'])
# cm → Variable storing the confusion matrix
# metrics.confusion_matrix → Function that shows counts of correct/incorrect predictions per class
# y_test, pred → Actual labels vs predicted labels being compared
# labels=['FAKE', 'REAL'] → Specifies the order of classes in the matrix

# This cell trains a Naive Bayes classifier on hashed features, predicts on test data, prints accuracy, and builds a confusion matrix 
# comparing FAKE vs REAL predictions.
```

```python
#13.GENERATING CLASSIFICATION REPORT

from sklearn.metrics import classification_report
# from → Python keyword used to import specific parts from a library
# sklearn.metrics → Module in scikit-learn containing evaluation metric functions
# import → Brings the specified function into the current program
# classification_report → Function that summarizes precision, recall, F1-score, and support for each class

report = classification_report(y_test, pred)
# report → Variable storing the classification report
# classification_report → Compares actual vs predicted labels and generates the summary
# y_test, pred → Actual labels vs predicted labels being compared

# This cell generates a detailed classification report showing precision, recall, and F1-score for each class.
```

```python
#14.VIEWING THE CLASSIFICATION REPORT

print(report)
# print → Built-in function that displays output
# report → Displays the classification report (precision, recall, F1-score, support for each class)

# This cell prints the classification report to review the model's performance in detail.
```
